## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# 1. Drop Duplicate Rows
titles_df.drop_duplicates(inplace=True)
credits_df.drop_duplicates(inplace=True)

# 2. Handle missing values in 'titles_df'
# Movies don't have seasons, so we fill the null 'seasons' values with 0
titles_df['seasons'].fillna(0, inplace=True)

# Fill missing text/categorical data with appropriate placeholders
titles_df['age_certification'].fillna('Not Rated', inplace=True)
titles_df['description'].fillna('No description available', inplace=True)

# 3. Handle missing values in 'credits_df'
# Not all roles (like directors) have character names, so we fill them with 'Unknown'
credits_df['character'].fillna('Unknown', inplace=True)

# 4. Quick check to verify our cleaning worked
print("Duplicates left in Titles dataset:", titles_df.duplicated().sum())
print("Duplicates left in Credits dataset:", credits_df.duplicated().sum())
print("\nMissing values in 'seasons' column:", titles_df['seasons'].isnull().sum())
print("Missing values in 'character' column:", credits_df['character'].isnull().sum())
````

### What all manipulations have you done and insights you found?

Data Manipulations & Insights:

To make the datasets analysis-ready, I performed the following data wrangling steps:

Removed Duplicates: I dropped the 3 duplicate rows from the titles dataset and the 56 duplicate rows from the credits dataset to ensure accuracy in our final visualizations.

Imputed Missing 'Seasons': The seasons column in the titles dataset had a massive amount of missing values (8,514). Because movies naturally do not have seasons, these weren't actually "missing" data points. I filled these nulls with 0 so the data remains intact for numerical analysis.

Categorical Imputation:

I filled missing age_certification values with the label 'Not Rated', as many older or independent titles on streaming platforms bypass official rating boards.

Missing description values were filled with 'No description available'.

In the credits dataset, I filled the 16,000+ missing character names with 'Unknown'. This accounts for crew members (like Directors) who do not portray characters, as well as uncredited background extras.

Preserved Rating Nuance: I intentionally left the NaN values in the rating columns (imdb_score, tmdb_score, etc.) as they are. Filling them with arbitrary means or medians could artificially skew the audience perception data. We will handle these specifically by filtering out nulls when plotting rating-based charts.

## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

#### Chart - 1

````python
# Chart - 1 visualization code: Distribution of Movies vs. TV Shows
plt.figure(figsize=(8, 6))

# Count the occurrences of each type
type_counts = titles_df['type'].value_counts()

# Create a pie chart
plt.pie(type_counts,
        labels=type_counts.index,
        autopct='%1.1f%%',
        startangle=90,
        colors=['#1f77b4', '#ff7f0e'],
        explode=(0.05, 0), # Slightly pull out the first slice for emphasis
        shadow=True)

plt.title('Distribution of Content Types: Movies vs. TV Shows', fontsize=15, fontweight='bold')
plt.show()
````

##### 1. Why did you pick the specific chart?

A pie chart is the most effective visualization for showing part-to-whole relationships. Because we are looking at a binary categorical variable (type: Movie vs. Show), the pie chart allows us to instantly visualize the proportion and market share of each content type within Amazon Prime's library.

##### 2. What is/are the insight(s) found from the chart?

The insight is striking: Movies absolutely dominate the Amazon Prime library. The platform is heavily skewed towards feature-length films, which make up roughly three-quarters of the entire US catalog, leaving TV Shows as a significantly smaller minority.

#### Chart - 2

````python
# Chart - 2 visualization code: Distribution of IMDb Scores
plt.figure(figsize=(10, 6))

# We drop nulls temporarily just for this visualization to avoid plotting errors
sns.histplot(titles_df['imdb_score'].dropna(), bins=40, kde=True, color='teal')

# Adding lines and labels for clarity
plt.axvline(titles_df['imdb_score'].mean(), color='red', linestyle='dashed', linewidth=2, label=f"Mean: {titles_df['imdb_score'].mean():.2f}")
plt.title('Distribution of IMDb Scores on Amazon Prime', fontsize=15, fontweight='bold')
plt.xlabel('IMDb Score', fontsize=12)
plt.ylabel('Number of Titles', fontsize=12)
plt.legend()

plt.show()
````

##### 1. Why did you pick the specific chart?

A histogram combined with a Kernel Density Estimate (KDE) line is the most effective chart for visualizing the distribution of a continuous numerical variable like IMDb scores. It allows us to instantly see where the bulk of the ratings cluster, the spread of the data, and the exact average (mean) rating of the platform's catalog.

##### 2. What is/are the insight(s) found from the chart?

The distribution of IMDb scores follows a roughly normal distribution, peaking slightly to the right of the center. The vast majority of Amazon Prime's content is rated between 5.5 and 7.5, with the mean score sitting at approximately 6.02. This indicates that while the platform has very few critically terrible titles, it also has a limited number of exceptional, highly-rated masterpieces (8.5+).

Are there any insights that lead to negative growth? Justify with specific reason.

Positive Impact: Yes. This data proves to stakeholders that the baseline quality of the catalog is acceptable (average to good).

#### Chart - 3

````python
# Chart - 3 visualization code: Top 10 Most Common Genres
plt.figure(figsize=(12, 6))

# Clean and split the genres column, then count occurrences
genres_series = titles_df['genres'].str.strip("[]").str.replace("'", "").str.split(", ")
all_genres = [genre.strip() for sublist in genres_series.dropna() for genre in sublist]

# Get the top 10 most frequent genres
top_genres = pd.Series(all_genres).value_counts().head(10)

# Create a horizontal bar plot for clear label reading
sns.barplot(x=top_genres.values, y=top_genres.index, palette='mako')

plt.title('Top 10 Most Common Genres on Amazon Prime', fontsize=15, fontweight='bold')
plt.xlabel('Number of Titles', fontsize=12)
plt.ylabel('Genre', fontsize=12)

plt.show()
````

##### 1. Why did you pick the specific chart?

A horizontal bar chart is the ideal visualization for ranking categorical variables with text labels (genres). Sorting the bars in descending order makes it exceptionally easy to compare frequency volumes without crowding or overlapping text along the axes.

##### 2. What is/are the insight(s) found from the chart?

Drama, Comedy, and Action are overwhelmingly the top three most prominent genres on Amazon Prime. Drama leads by a significant margin, followed closely by Comedy, proving that these two broad categories form the core foundation of the platform's content library.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

#### Chart - 4

````python
# Chart - 4 visualization code: Content Release Trend Over the Years
plt.figure(figsize=(12, 6))

# Filter out extremely old outliers for a cleaner modern trend view (e.g., from year 2000 onwards)
recent_releases = titles_df[titles_df['release_year'] >= 2000]

# Group by release year and count titles, splitting by type
yearly_trend = recent_releases.groupby(['release_year', 'type']).size().unstack().fillna(0)

# Create a line plot
yearly_trend.plot(kind='line', marker='o', figsize=(12, 6), linewidth=2.5)

plt.title('Trend of Content Releases on Amazon Prime (2000 - Present)', fontsize=15, fontweight='bold')
plt.xlabel('Release Year', fontsize=12)
plt.ylabel('Number of Titles Released', fontsize=12)
plt.legend(title='Content Type')
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
````

##### 1. Why did you pick the specific chart?

A multi-line chart is the industry standard for visualizing continuous time-series trends. By tracking the release years on the x-axis and the volume of titles on the y-axis, while separating movies and TV shows into distinct lines, we can easily compare growth trajectories and historical production spikes over time.

##### 2. What is/are the insight(s) found from the chart?

The chart reveals an exponential upward spike in content production starting around the mid-2010s, reflecting Amazon's aggressive expansion of its Prime Video catalog. While movie releases grew massively, TV show production also saw a steady, parallel acceleration, highlighting the platform's rising investment in serial content.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Positive Impact: Yes. Understanding the historical velocity of content additions helps financial planners align future capital expenditures with historical growth curves.
Negative Growth Risk: A sudden downward tail at the very end of the timeline (representing recent years) might appear due to dataset lag or incomplete reporting of upcoming content. #### Chart - 5

````python
# Chart - 5 visualization code: Correlation between TMDB Popularity and IMDb Score
plt.figure(figsize=(10, 6))

# Drop nulls for accurate scatter plotting
clean_ratings = titles_df.dropna(subset=['tmdb_popularity', 'imdb_score'])

# Create a scatter plot with a regression trendline
sns.scatterplot(data=clean_ratings, x='tmdb_popularity', y='imdb_score', alpha=0.5, color='purple')

plt.title('IMDb Score vs. TMDB Popularity', fontsize=15, fontweight='bold')
plt.xlabel('TMDB Popularity Score', fontsize=12)
plt.ylabel('IMDb Score', fontsize=12)

plt.show()
````

````python
# Chart - 6 visualization code: Runtime Distribution by Content Type
plt.figure(figsize=(10, 6))

# Filter out extreme runtime outliers for clean visualization
clean_runtime = titles_df[(titles_df['runtime'] > 0) & (titles_df['runtime'] < 300)]

sns.boxplot(data=clean_runtime, x='type', y='runtime', palette='Set2')

plt.title('Runtime Distribution: Movies vs. TV Shows', fontsize=15, fontweight='bold')
plt.xlabel('Content Type', fontsize=12)
plt.ylabel('Runtime (Minutes)', fontsize=12)

plt.show()
````
